```python
import sys
import os

# Find the project root (Speciale_Kode)
current_dir = os.getcwd()
project_root = current_dir

# Looks for "Speciale_Kode" folder:
while os.path.basename(project_root) != "Speciale_Kode":
    project_root = os.path.dirname(project_root)

# Add to Python path
if project_root not in sys.path:
    sys.path.append(project_root)
```

```python
baseline_folder = os.path.join(project_root, "Baseline")

naive_folder = os.path.join(baseline_folder, "Naive")
MA_folder = os.path.join(baseline_folder, "MovingAverage")
seasonal_folder = os.path.join(baseline_folder, "Seasonal")
```

```python
import pandas as pd

# Make averages for each day of 7 day forecast period for Naive Persistence

filename_naive_DK1 = "DK1_weekly_results_naive.csv"
filepath_naive_DK1 = os.path.join(naive_folder, filename_naive_DK1)

df_naive_DK1 = pd.read_csv(filepath_naive_DK1)

naive_DK1_averages = {}

column = "weekly_smape"
for i in range(7):
    naive_DK1_averages[f'avg_smape_day_{i+1}'] = df_naive_DK1[column].iloc[i::7].mean()

naive_DK1 = list(naive_DK1_averages.values())

filename_naive_DK2 = "DK2_weekly_results_naive.csv"
filepath_naive_DK2 = os.path.join(naive_folder, filename_naive_DK2)

df_naive_DK2 = pd.read_csv(filepath_naive_DK2)

naive_DK2_averages = {}

column = "weekly_smape"
for i in range(7):
    naive_DK2_averages[f'avg_smape_day_{i+1}'] = df_naive_DK2[column].iloc[i::7].mean()

naive_DK2 = list(naive_DK2_averages.values())
```

```python
# Make averages for each day of 7 day forecast period for Moving Average

filename_MA_DK1 = "DK1_weekly_results_MA.csv"
filepath_MA_DK1 = os.path.join(MA_folder, filename_MA_DK1)

df_MA_DK1 = pd.read_csv(filepath_MA_DK1)

MA_DK1_averages = {}

column = "weekly_smape"
for i in range(7):
    MA_DK1_averages[f'avg_smape_day_{i+1}'] = df_MA_DK1[column].iloc[i::7].mean()

MovingA_DK1 = list(MA_DK1_averages.values())


filename_MA_DK2 = "DK2_weekly_results_MA.csv"
filepath_MA_DK2 = os.path.join(MA_folder, filename_MA_DK2)

df_MA_DK2 = pd.read_csv(filepath_MA_DK2)

MA_DK2_averages = {}

column = "weekly_smape"
for i in range(7):
    MA_DK2_averages[f'avg_smape_day_{i+1}'] = df_MA_DK2[column].iloc[i::7].mean()

MovingA_DK2 = list(MA_DK2_averages.values())
```

```python
# Make averages for each day of 7 day forecast period for Moving Average

filename_seasonal_DK1 = "DK1_weekly_results_seasonal.csv"
filepath_seasonal_DK1 = os.path.join(seasonal_folder, filename_seasonal_DK1)

df_seasonal_DK1 = pd.read_csv(filepath_seasonal_DK1)

seasonal_DK1_averages = {}

column = "weekly_smape"
for i in range(7):
    seasonal_DK1_averages[f'avg_smape_day_{i+1}'] = df_seasonal_DK1[column].iloc[i::7].mean()

seasonal_DK1 = list(seasonal_DK1_averages.values())


filename_seasonal_DK2 = "DK2_weekly_results_seasonal.csv"
filepath_seasonal_DK2 = os.path.join(seasonal_folder, filename_seasonal_DK2)

df_seasonal_DK2 = pd.read_csv(filepath_seasonal_DK2)

seasonal_DK2_averages = {}

column = "weekly_smape"
for i in range(7):
    seasonal_DK2_averages[f'avg_smape_day_{i+1}'] = df_seasonal_DK2[column].iloc[i::7].mean()

seasonal_DK2 = list(seasonal_DK2_averages.values())
```

```python
import matplotlib.pyplot as plt

# Generating a plot for baseline DK1 7-Day 

xpoints = ["Day 1", "Day 2", "Day 3", "Day 4", "Day 5", "Day 6", "Day 7"]

plt.plot(xpoints, naive_DK1, label = "Naive Persistence")
plt.plot(xpoints, MovingA_DK1, label = "Moving Average")
plt.plot(xpoints, seasonal_DK1, label = "Seasonal Naive")
plt.xlabel('Prediction Day')
plt.ylabel('SMAPE (%)')
plt.title('7-Day SMAPE (%) Development for Baseline Models for DK1')
plt.legend()
plt.show()
```

```python
# Generating a plot for baseline DK1 7-Day 

xpoints = ["Day 1", "Day 2", "Day 3", "Day 4", "Day 5", "Day 6", "Day 7"]

plt.plot(xpoints, naive_DK2, label = "Naive Persistence")
plt.plot(xpoints, MovingA_DK2, label = "Moving Average")
plt.plot(xpoints, seasonal_DK2, label = "Seasonal Naive")
plt.xlabel('Prediction Day')
plt.ylabel('SMAPE (%)')
plt.title('7-Day SMAPE (%) Development for Baseline Models for DK2')
plt.legend()
plt.show()
```
