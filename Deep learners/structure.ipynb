## Neural network structure ideas

LSTM enocder/decoder:
link: https://github.com/lkulowski/LSTM_encoder_decoder/tree/master

```python
class TabularLSTM(nn.Module):
    def __init__(self, input_size, hidden_size, layers, dropout=0.0):
        super().__init__()
        self.lstm = nn.LSTM(
            input_size=input_size,
            hidden_size=hidden_size,
            num_layers=layers,
            dropout=dropout if layers > 1 else 0.0,
            batch_first=True,
        )
        self.fc = nn.Linear(hidden_size, 1)

    def forward(self, x):
        out, _ = self.lstm(x)         
        return self.fc(out[:, -1, :])  
```

```python
class TabularGRU(nn.Module):
    def __init__(self, input_size, hidden_size, layers, dropout=0.0):
        super().__init__()
        self.gru = nn.GRU(
            input_size=input_size,
            hidden_size=hidden_size,
            num_layers=layers,
            dropout=dropout if layers > 1 else 0.0,
            batch_first=True,
        )
        self.fc = nn.Linear(hidden_size, 1)

    def forward(self, x):
        out, _ = self.gru(x)           
        return self.fc(out[:, -1, :])
```
